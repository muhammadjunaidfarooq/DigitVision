```python
import numpy as np
import matplotlib.pyplot as plt

# Create a small grayscale image using pixel values.
image = np.array([
    [0,   0, 255,   0,   0],
    [0, 255, 255,   0,   0],
    [0,   0, 255,   0,   0],
    [0,   0, 255,   0,   0],
    [0, 255, 255, 255,   0]
], dtype=np.uint8)

# Display the image as a grayscale grid.
plt.imshow(
    image,
    cmap="gray",
    vmin=0,
    vmax=255,
    interpolation="nearest"
)

plt.title("Our First 5 × 5 Image")
plt.colorbar(label="Pixel intensity")
plt.show()

# Inspect its numerical representation.
print("Image shape:", image.shape)
print("Total pixels:", image.size)
print("Pixel at row 0, column 2:", image[0, 2])
```
